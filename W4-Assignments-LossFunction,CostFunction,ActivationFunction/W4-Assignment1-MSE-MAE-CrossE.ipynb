```python
'''1) Implement the following Mean Squared Error (MSE) loss function in Python 
 
Given the true values 𝑦 = [0.000, 0.166, 0.333], and predicted values 𝑦 = [0.000, 0.254, 
0.998], use the implemented function to calculate error.''' 
```

```python
import numpy as np
y_hat= np.array([0.000, 0.166, 0.333])
y_true= np.array([0.000, 0.254, 0.998])

def rmse(predictions, targets):
    differences = predictions - targets
    differences_squared = differences ** 2
    mean_of_differences_squared = differences_squared.mean()
    rmse_val = np.sqrt(mean_of_differences_squared)
    return rmse_val

print("predicted: " + str(["%.8f" % elem for elem in y_hat])) 
print("True is: " + str(["%.8f" % elem for elem in y_true]))
rmse_val =  rmse(y_hat, y_true)
print("rms error is: " + str(rmse_val))
```

```python
'''3) Implement the following Cross Entropy loss function in Python 
 
Given the true values 𝑦 = [[0,0,0,1], [0,0,0,1]],  
predicted values 𝑦 = [[0.25,0.25,0.25,0.25], [0.01,0.01,0.01,0.96]] and k is the number of 
classes, use the implemented function to calculate error.'''
```

```python
import numpy as np 
predictions = np.array([[0.25,0.25,0.25,0.25], [0.01,0.01,0.01,0.96]]) 
targets = np.array([[0,0,0,1], [0,0,0,1]]) 
def cross_entropy(predictions, targets, epsilon=1e-10): 
    predictions = np.clip(predictions, epsilon, 1. - epsilon) #clip() function is used to Clip (limit) the values in an array
    N = predictions.shape[0] #NumPy arrays have an attribute called shape that returns a tuple with each index having the number of corresponding elements. 
    print(N)
    ce_loss = -np.sum(np.sum(targets * np.log(predictions + 1e-5)))/N #This function returns the sum of array elements over the specified axis
    return ce_loss 
cross_entropy_loss = cross_entropy(predictions, targets) 
print ("Cross entropy loss is: " + str(cross_entropy_loss)) 
```
